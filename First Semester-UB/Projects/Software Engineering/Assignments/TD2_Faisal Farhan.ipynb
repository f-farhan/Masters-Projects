```python
while True:
    try:
        user_input = input("Enter your integer number: ")
        number = int(user_input)
        break
    except ValueError:
        print("Input Value Error")

print("Number entered:", number)
```

# Classes

Create a class Complex that could be used according to:

```python
z1 = Complex (3 , 4)
z1.real
# 3
z1.imag
# 4
z1.mod ()
# 5.0
z1.arg ()
# 0.9272952180016122
z1.conj().imag
# -4
```

Add the method polar to print a polar representation of the complex number:

```python
z1 . polar ()
#'5.0* exp (0.92779 i) '
```

Solution

```python
import math

class Complex:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def mod(self):
        return math.sqrt(self.real**2 + self.imag**2)

    def arg(self):
        return math.atan2(self.imag, self.real)

    def conj(self):
        return Complex(self.real, -self.imag)
    def polar(self):
        modulus = self.mod()
        argument = self.arg()
        print(f"Modulus: {modulus}, Argument: {argument}")

# Usage example
z1 = Complex(3, 4)
print("Real value: ", z1.real)
print("Imaginary value: ",z1.imag)
print("Modulus: ",z1.mod())
print("Argument: ",z1.arg())
print("Conjugatee: ",z1.conj().imag)
z1.polar()
```
